**1 питання-гіпотеза.**
Якщо в Naypyitaw спостерігається найбільший середній чек та валовий дохід, хоча найменша кількість покупок, то хто ці люди що роблять ці покупки (який тип клієнта Customer type) - чи дійсно найвийщий середній чек роблять клінти типу Member? Потрібно проаналізувати чи відрізняється клієнти за типом (Member / Normal) і їхній середній чек у Naypyitaw порівняно з іншими містами?:
*   середній чек за типом клієнта в кожному місті
*   частка покупок кожного типу клієнта в кожному місті

```python
total_mean = df.groupby(['City', 'Customer type'])['Total'].mean().round(2)
```

```python
total_mean
```

```python
perc_of_customers=df.groupby('City')['Customer type'].value_counts(normalize=True)*100
perc_of_customers
```

```python
pt_total_mean=df.pivot_table(index='City', columns='Customer type', values='Total', aggfunc='mean').round(2)
pt_total_mean
```

```python
pt=df.pivot_table(index='City', columns='Customer type', values='Total', aggfunc='count')
```

```python
pt.sum(axis=1)
```

```python
pt_perc_of_customers=(pt.div(pt.sum(axis=1), axis=0)*100).round(2)
pt_perc_of_customers
```

```python
final_pivot = pd.concat([pt_total_mean, pt_perc_of_customers], keys=['Середній чек', 'Частка покупок, %'], axis=1)
final_pivot
```

```python
normal = final_pivot['Середній чек']['Normal']
member = final_pivot['Середній чек']['Member']

print((normal.max() / normal - 1) * 100)
print((member.max() / member - 1) * 100)
```

Отже, тип клієнта Normal має вищий середній чек у Naypyitaw аніж тип Member - гіпотеза не підтвердилася, в цьому місті клієнти Member не роблять вищий середній чек. Та інші спостереження показують, що в інших двох містах найвищий середній чек є серед клієнтів типу Member. Щодо кількості покупок, то різниця між типами Member та Normal у кожному з міст є несуттєвою у відсотковому співвідношенні:
у м. Naypyitaw покупці типу Member здійснють більше покупок на 3 п.п, на відмінну від типу Normal. в м. Mandalay та Yangon - покупці типу Normal всього на 0.6 і 1.8 п.п більше здійснюють покупок більше аніж Member.
По середньому чеку Normal випереджає Member у Naypyitaw на 0.3%, тобто чек практично однаковий. А в інших містах Member випереджають Normal помітніше (Mandalay на 3.5% та Yangon на 5.7%).
Тип клієнта Normal по середньому чеку у Naypyitaw випереджає Normal в інших містах (порядку 7-11%), а тип Member на 3-5%.
Чек у Naypyitaw вищий в обох типів клієнтів, тож високий чек у Naypyitaw не пояснюється типом клієнта, причину варто шукати в інших факторах (асортимент, ціни тощо).

**3 питання-гіпотеза**. Чи відрізняються кількість покупок і середній чек у будні та у вихідні, і чи однаково це у всіх містах? Гіпотеза - у м. Naypyitaw найчастіше і найбільше купують в робочі дні.




```python
df['Date_dt'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')
df['Weekday'] = df['Date_dt'].dt.day_name()
df[['Date', 'Weekday']].head()
```

```python
df['Is_weekend'] = df['Date_dt'].dt.dayofweek >= 5
```

```python
pt_day_mean=df.pivot_table(index='City', columns='Is_weekend', values='Total', aggfunc='mean').round(2)
pt_day_mean
```

```python
pt_day_count=df.pivot_table(index='City', columns='Is_weekend', values='Total', aggfunc='count').round(2)
pt_day_count
```

```python
pt_day_mean.idxmax(axis=1)
```

```python
pt_day_count.idxmax(axis=1)
```

```python
pt_we_per_day = pt_day_count.div([5, 2], axis=1).round(1)
print(pt_day_mean)
print(pt_we_per_day)
```

Гіпотеза не підтвердилась: у Naypyitaw у вихідні і середній чек, і кількість покупок за день вищі, ніж у будні. Середній чек становить 351.07 проти 330.97 (+6.1%), покупок на день 50.0 проти 45.6 (+9.6%). Така сама тенденція за чеком у всіх містах: у вихідні він вищий на 6.2% у Mandalay, 6.1% у Naypyitaw і 8.8% у Yangon. За кількістю покупок на день у Yangon вихідні вищі на 7.1%, а в Mandalay різниці немає (47.5 і 47.4).
